```python
# ============================================================
# CELL 6: SALES DATA QUALITY ANALYSIS
# Purpose:
# 1. Check missing sales values.
# 2. Check negative sales values.
# 3. Calculate zero-sales frequency.
# 4. Check duplicate product-store records.
# ============================================================

# Check missing values in daily sales columns
missing_counts = sales_df.select([
    pl.col(col).is_null().sum().alias(col)
    for col in day_columns
])

total_missing = sum(missing_counts.row(0))

# Count zero and negative sales
zero_counts = sales_df.select([
    (pl.col(col) == 0).sum().alias(col)
    for col in day_columns
])

negative_counts = sales_df.select([
    (pl.col(col) < 0).sum().alias(col)
    for col in day_columns
])

total_zero_sales = sum(zero_counts.row(0))
total_negative_sales = sum(negative_counts.row(0))

total_observations = sales_df.height * len(day_columns)

# Check duplicate product-store combinations
duplicate_records = (
    sales_df.height - sales_df["id"].n_unique()
)

print("SALES DATA QUALITY REPORT")
print("=" * 45)

print("Total sales observations:", total_observations)
print("Missing sales values:", total_missing)

print(
    "Zero sales observations:",
    total_zero_sales,
    f"({total_zero_sales / total_observations:.2%})"
)

print("Negative sales observations:", total_negative_sales)
print("Duplicate product-store records:", duplicate_records)

print("\nSALES QUALITY CHECK COMPLETED!")
```

```python
# ============================================================
# CELL 9: WIDE-TO-LONG PANEL TRANSFORMATION
# Purpose:
# 1. Convert daily sales columns into rows using Polars.
# 2. Merge calendar information.
# 3. Merge weekly selling prices.
# 4. Generate state-specific SNAP flags.
# 5. Save the complete panel dataset to Parquet.
# ============================================================

# Define metadata columns
metadata_columns = [
    "item_id",
    "store_id",
    "state_id"
]

# Step 1: Convert sales data into a lazy frame
sales_lazy = sales_df.lazy()

# Step 2: Convert wide sales data into long format
sales_long = sales_lazy.unpivot(
    on=day_columns,
    index=metadata_columns,
    variable_name="d",
    value_name="sales"
)

# Step 3: Prepare calendar information
calendar_features = (
    calendar_df.lazy()
    .select([
        "d",
        "date",
        "wm_yr_wk",
        "event_name_1",
        "snap_CA",
        "snap_TX",
        "snap_WI"
    ])
    .with_columns(
        pl.col("date").str.to_date("%Y-%m-%d")
    )
)

# Step 4: Merge calendar information
panel_lazy = sales_long.join(
    calendar_features,
    on="d",
    how="left"
)

# Step 5: Generate state-specific SNAP flag
panel_lazy = panel_lazy.with_columns(
    pl.when(pl.col("state_id") == "CA")
    .then(pl.col("snap_CA"))
    .when(pl.col("state_id") == "TX")
    .then(pl.col("snap_TX"))
    .when(pl.col("state_id") == "WI")
    .then(pl.col("snap_WI"))
    .otherwise(None)
    .alias("snap_flag")
)

# Step 6: Prepare weekly price information
prices_lazy = prices_df.lazy().select([
    "store_id",
    "item_id",
    "wm_yr_wk",
    "sell_price"
])

# Step 7: Merge selling prices
panel_lazy = panel_lazy.join(
    prices_lazy,
    on=["store_id", "item_id", "wm_yr_wk"],
    how="left"
)

# Step 8: Select required panel columns
panel_lazy = panel_lazy.select([
    "item_id",
    "store_id",
    "date",
    "sales",
    "sell_price",
    "event_name_1",
    "snap_flag"
])

# Step 9: Save using streaming execution
PANEL_PATH = PROCESSED_DATA_DIR / "m5_panel.parquet"

panel_lazy.sink_parquet(
    PANEL_PATH,
    compression="zstd",
    engine="streaming"
)

print("PANEL TRANSFORMATION COMPLETED!")
print("Saved location:", PANEL_PATH)
```

```python

# ============================================================
# CELL 11: CREATE 3-FOLD ROLLING-ORIGIN BACKTESTING SPLITS
# Purpose:
# 1. Find the first and last dates in the panel.
# 2. Create three consecutive 28-day validation windows.
# 3. Define an expanding training period for each fold.
# 4. Verify that every validation window has exactly 28 days.
# ============================================================

from datetime import timedelta
import pandas as pd

# Step 1: Get the first and last dates from the panel
date_limits = panel_df.select([
    pl.col("date").min().alias("first_date"),
    pl.col("date").max().alias("last_date")
]).collect()

first_date = date_limits["first_date"][0]
last_date = date_limits["last_date"][0]

print("FIRST PANEL DATE:", first_date)
print("LAST PANEL DATE:", last_date)

# Step 2: Define the forecast horizon
HORIZON = 28
NUMBER_OF_FOLDS = 3

# Step 3: Calculate the beginning of the three validation windows
last_validation_start = last_date - timedelta(days=HORIZON - 1)
first_validation_start = last_validation_start - timedelta(
    days=HORIZON * (NUMBER_OF_FOLDS - 1)
)

# Step 4: Create the fold definitions
folds = []

for fold_number in range(1, NUMBER_OF_FOLDS + 1):

    # Calculate the validation start and end dates
    validation_start = first_validation_start + timedelta(
        days=HORIZON * (fold_number - 1)
    )

    validation_end = validation_start + timedelta(days=HORIZON - 1)

    # Training data ends immediately before validation starts
    training_end = validation_start - timedelta(days=1)

    folds.append({
        "fold": fold_number,
        "training_start": first_date,
        "training_end": training_end,
        "validation_start": validation_start,
        "validation_end": validation_end
    })

# Step 5: Display the split definitions
fold_table = pd.DataFrame(folds)

print("\nROLLING-ORIGIN BACKTESTING SPLITS")
display(fold_table)

# Step 6: Verify each validation window contains exactly 28 days
for fold in folds:

    validation_days = (
        fold["validation_end"] - fold["validation_start"]
    ).days + 1

    assert validation_days == HORIZON

    print(
        f"Fold {fold['fold']}: "
        f"{validation_days} validation days | "
        f"{fold['validation_start']} to {fold['validation_end']}"
    )

print("\nBACKTESTING SPLIT DEFINITIONS CREATED SUCCESSFULLY!")
```

```python

# ============================================================
# CELL 12: CREATE TRAINING AND VALIDATION DATASETS
# Purpose:
# 1. Create separate lazy datasets for each backtesting fold.
# 2. Keep training and validation periods separate.
# 3. Count observations per date using one grouped scan.
# 4. Verify the number of days and rows in every fold.
# ============================================================

# Step 1: Create lazy training and validation datasets
fold_splits = {}

for fold in folds:

    fold_number = fold["fold"]
    training_start = fold["training_start"]
    training_end = fold["training_end"]
    validation_start = fold["validation_start"]
    validation_end = fold["validation_end"]

    # Training data: all dates before the validation window
    train_data = panel_df.filter(
        pl.col("date").is_between(
            training_start,
            training_end,
            closed="both"
        )
    )

    # Validation data: exactly 28 future days
    validation_data = panel_df.filter(
        pl.col("date").is_between(
            validation_start,
            validation_end,
            closed="both"
        )
    )

    # Store the lazy datasets for later model training
    fold_splits[fold_number] = {
        "train": train_data,
        "validation": validation_data
    }

# Step 2: Count observations for each date
# This scans the panel once and returns a small daily summary.
daily_counts = (
    panel_df
    .group_by("date")
    .agg(pl.len().alias("rows"))
    .sort("date")
    .collect()
)

# Step 3: Verify training and validation sizes
EXPECTED_SERIES_PER_DAY = 30490

for fold in folds:

    fold_number = fold["fold"]

    # Calculate expected training and validation days
    training_days_expected = (
        fold["training_end"] - fold["training_start"]
    ).days + 1

    validation_days_expected = HORIZON

    # Count actual training observations and days
    train_summary = (
        daily_counts
        .filter(
            pl.col("date").is_between(
                fold["training_start"],
                fold["training_end"],
                closed="both"
            )
        )
        .select([
            pl.len().alias("days"),
            pl.col("rows").sum().alias("rows")
        ])
        .row(0, named=True)
    )

    # Count actual validation observations and days
    validation_summary = (
        daily_counts
        .filter(
            pl.col("date").is_between(
                fold["validation_start"],
                fold["validation_end"],
                closed="both"
            )
        )
        .select([
            pl.len().alias("days"),
            pl.col("rows").sum().alias("rows")
        ])
        .row(0, named=True)
    )

    # Calculate expected row counts
    expected_train_rows = (
        training_days_expected * EXPECTED_SERIES_PER_DAY
    )

    expected_validation_rows = (
        validation_days_expected * EXPECTED_SERIES_PER_DAY
    )

    # Verify dates and row counts
    assert train_summary["days"] == training_days_expected
    assert validation_summary["days"] == validation_days_expected

    assert train_summary["rows"] == expected_train_rows
    assert validation_summary["rows"] == expected_validation_rows

    print(f"\nFOLD {fold_number}")
    print("Training days:", train_summary["days"])
    print("Training rows:", train_summary["rows"])
    print("Validation days:", validation_summary["days"])
    print("Validation rows:", validation_summary["rows"])

print("\nALL THREE FOLD DATASETS VERIFIED SUCCESSFULLY!")
```

```python

# ============================================================
# CELL 13: SEASONAL NAIVE BASELINE MODEL
# Purpose:
# 1. Use the previous 7 days of sales as the weekly pattern.
# 2. Repeat that pattern over the next 28 days.
# 3. Generate predictions for all product-store combinations.
# 4. Evaluate predictions using MAE, RMSE, and WAPE.
# 5. Save predictions for all three backtesting folds.
# ============================================================

import numpy as np
from datetime import timedelta

# Create a dictionary to store predictions for each fold
seasonal_naive_results = {}

# Store evaluation metrics for all folds
seasonal_naive_metrics = []

# Create the forecast horizon table (1 to 28 days)
horizon_table = pl.DataFrame({
    "horizon": list(range(1, HORIZON + 1))
})

# Create the predictions folder if it does not exist
PREDICTION_DIR.mkdir(parents=True, exist_ok=True)

# Run Seasonal Naive for each backtesting fold
for fold in folds:

    fold_number = fold["fold"]
    training_end = fold["training_end"]
    validation_start = fold["validation_start"]

    print(f"\nPROCESSING SEASONAL NAIVE - FOLD {fold_number}")

    # Step 1: Extract the last 7 days of training sales
    last_week_start = training_end - timedelta(days=6)

    last_week = (
        fold_splits[fold_number]["train"]
        .filter(
            pl.col("date").is_between(
                last_week_start,
                training_end,
                closed="both"
            )
        )
        .select([
            "item_id",
            "store_id",
            "date",
            "sales"
        ])
        .collect()
    )

    # Step 2: Generate forecasts for all 28 horizons
    # Each historical weekday is matched with the same
    # weekday in the future forecast period.
    forecasts = (
        last_week
        .join(horizon_table, how="cross")
        .filter(
            pl.col("date").dt.weekday()
            == ((pl.col("horizon") - 1) % 7 + 1)
        )
        .with_columns([
            (
                pl.lit(validation_start)
                + pl.duration(days=pl.col("horizon") - 1)
            ).alias("forecast_date"),
            pl.col("sales").alias("forecast")
        ])
        .select([
            "item_id",
            "store_id",
            "forecast_date",
            "horizon",
            "forecast"
        ])
    )

    # Step 3: Load actual validation sales
    actuals = (
        fold_splits[fold_number]["validation"]
        .select([
            "item_id",
            "store_id",
            pl.col("date").alias("forecast_date"),
            pl.col("sales").alias("actual")
        ])
        .collect()
    )

    # Step 4: Match forecasts with actual sales
    scored = forecasts.join(
        actuals,
        on=["item_id", "store_id", "forecast_date"],
        how="inner"
    )

    # Step 5: Verify prediction row count
    prediction_rows = scored.height

    expected_rows = EXPECTED_SERIES_PER_DAY * HORIZON

    assert prediction_rows == expected_rows

    # Step 6: Calculate evaluation metrics
    metric_values = scored.select([
        (pl.col("actual") - pl.col("forecast"))
        .abs()
        .mean()
        .alias("MAE"),

        (
            (
                pl.col("actual") - pl.col("forecast")
            ).pow(2).mean().sqrt()
        ).alias("RMSE"),

        (
            (
                pl.col("actual") - pl.col("forecast")
            ).abs().sum()
            / pl.col("actual").sum()
            * 100
        ).alias("WAPE_Percent")
    ]).row(0, named=True)

    # Step 7: Save predictions
    prediction_path = (
        PREDICTION_DIR
        / f"seasonal_naive_fold_{fold_number}.parquet"
    )

    scored.write_parquet(
        prediction_path,
        compression="zstd"
    )

    # Store results
    seasonal_naive_results[fold_number] = scored

    seasonal_naive_metrics.append({
        "fold": fold_number,
        "prediction_rows": prediction_rows,
        **metric_values
    })

    # Display fold results
    print("Prediction rows:", prediction_rows)
    print("MAE:", round(metric_values["MAE"], 4))
    print("RMSE:", round(metric_values["RMSE"], 4))
    print("WAPE (%):", round(metric_values["WAPE_Percent"], 4))
    print("Saved:", prediction_path)

# Step 8: Display combined metrics
seasonal_naive_summary = pd.DataFrame(seasonal_naive_metrics)

print("\nSEASONAL NAIVE - FINAL BACKTEST RESULTS")
display(seasonal_naive_summary)

print("\nSEASONAL NAIVE IMPLEMENTATION COMPLETED!")
```

```python

# ============================================================
# CELL 14: CROSTON'S SBA BASELINE MODEL
# Purpose:
# 1. Estimate intermittent demand for each product-store series.
# 2. Estimate demand size and time between non-zero sales.
# 3. Apply the SBA bias correction.
# 4. Forecast all 28 validation days.
# 5. Evaluate and save predictions for all three folds.
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

# Step 1: Define Croston's SBA forecasting function
def croston_sba_forecast(history, alpha=0.1):

    # Convert the historical sales sequence to a list
    values = [float(value) for value in history]

    # Find the positions where sales were greater than zero
    nonzero_positions = [
        index for index, value in enumerate(values)
        if value > 0
    ]

    # If the product never sold, forecast zero
    if len(nonzero_positions) == 0:
        return 0.0

    # Initialize demand size and demand interval
    first_position = nonzero_positions[0]

    demand_size = values[first_position]
    demand_interval = float(first_position + 1)

    previous_position = first_position

    # Update estimates whenever a new sale occurs
    for position in nonzero_positions[1:]:

        current_demand = values[position]

        # Calculate the interval between two sales
        interval = position - previous_position

        # Exponentially smooth demand size
        demand_size = (
            demand_size
            + alpha * (current_demand - demand_size)
        )

        # Exponentially smooth demand interval
        demand_interval = (
            demand_interval
            + alpha * (interval - demand_interval)
        )

        previous_position = position

    # Apply the SBA bias correction
    forecast = (
        (1 - alpha / 2)
        * demand_size
        / demand_interval
    )

    return max(0.0, forecast)


# Step 2: Set the smoothing parameter
CROSTON_ALPHA = 0.1

# Step 3: Store results
croston_results = {}
croston_metrics = []

# Step 4: Run Croston for each backtesting fold
for fold in folds:

    fold_number = fold["fold"]
    validation_start = fold["validation_start"]

    print(f"\nPROCESSING CROSTON SBA - FOLD {fold_number}")

    # Step 5: Group historical sales by product and store.
    # Sort each series chronologically before forecasting.
    historical_series = (
        fold_splits[fold_number]["train"]
        .group_by(["item_id", "store_id"])
        .agg(
            pl.col("sales")
            .sort_by("date")
            .alias("sales_history")
        )
        .with_columns(
            pl.col("sales_history")
            .map_elements(
                lambda history: croston_sba_forecast(
                    history,
                    alpha=CROSTON_ALPHA
                ),
                return_dtype=pl.Float64
            )
            .alias("forecast")
        )
        .select([
            "item_id",
            "store_id",
            "forecast"
        ])
        .collect()
    )

    # Step 6: Verify one forecast for every product-store series
    assert historical_series.height == EXPECTED_SERIES_PER_DAY

    # Step 7: Expand each series forecast to 28 future days
    forecasts = (
        historical_series
        .join(horizon_table, how="cross")
        .with_columns(
            (
                pl.lit(validation_start)
                + pl.duration(days=pl.col("horizon") - 1)
            ).alias("forecast_date")
        )
        .select([
            "item_id",
            "store_id",
            "forecast_date",
            "horizon",
            "forecast"
        ])
    )

    # Step 8: Load actual validation sales
    actuals = (
        fold_splits[fold_number]["validation"]
        .select([
            "item_id",
            "store_id",
            pl.col("date").alias("forecast_date"),
            pl.col("sales").alias("actual")
        ])
        .collect()
    )

    # Step 9: Compare forecasts against actual sales
    scored = forecasts.join(
        actuals,
        on=["item_id", "store_id", "forecast_date"],
        how="inner"
    )

    # Step 10: Verify prediction count
    assert scored.height == EXPECTED_SERIES_PER_DAY * HORIZON

    # Step 11: Calculate forecasting metrics
    metric_values = scored.select([
        (
            (pl.col("actual") - pl.col("forecast"))
            .abs()
            .mean()
        ).alias("MAE"),

        (
            (
                pl.col("actual") - pl.col("forecast")
            ).pow(2).mean().sqrt()
        ).alias("RMSE"),

        (
            (
                (pl.col("actual") - pl.col("forecast"))
                .abs()
                .sum()
                / pl.col("actual").sum()
            ) * 100
        ).alias("WAPE_Percent")
    ]).row(0, named=True)

    # Step 12: Save predictions
    prediction_path = (
        PREDICTION_DIR
        / f"croston_sba_fold_{fold_number}.parquet"
    )

    scored.write_parquet(
        prediction_path,
        compression="zstd"
    )

    # Step 13: Store results
    croston_results[fold_number] = scored

    croston_metrics.append({
        "fold": fold_number,
        "prediction_rows": scored.height,
        **metric_values
    })

    # Step 14: Display results
    print("Product-store series:", historical_series.height)
    print("Prediction rows:", scored.height)
    print("MAE:", round(metric_values["MAE"], 4))
    print("RMSE:", round(metric_values["RMSE"], 4))
    print("WAPE (%):", round(metric_values["WAPE_Percent"], 4))
    print("Saved:", prediction_path)


# Step 15: Display final results
croston_summary = pd.DataFrame(croston_metrics)

print("\nCROSTON SBA - FINAL BACKTEST RESULTS")
display(croston_summary)

print("\nCROSTON SBA IMPLEMENTATION COMPLETED!")
```

```python

# ============================================================
# CELL 17: LEAK-FREE TEMPORAL FEATURE ENGINEERING
# Purpose:
# 1. Load the CA_1 store data.
# 2. Sort each product's sales chronologically.
# 3. Create lag features with a 28-day forecast gap.
# 4. Create rolling averages using only available historical sales.
# 5. Create calendar and price features.
# 6. Check missing values and feature availability.
# ============================================================

# Step 1: Load only the selected store
ca1_data = (
    pl.scan_parquet(PANEL_PATH)
    .filter(pl.col("store_id") == PILOT_STORE)
    .select([
        "item_id",
        "store_id",
        "date",
        "sales",
        "sell_price",
        "event_name_1",
        "snap_flag"
    ])
    .sort(["item_id", "date"])
)

# Step 2: Create lag features
ca1_features = ca1_data.with_columns([
    pl.col("sales").shift(28).over("item_id").alias("lag_28"),
    pl.col("sales").shift(35).over("item_id").alias("lag_35"),
    pl.col("sales").shift(42).over("item_id").alias("lag_42"),
    pl.col("sales").shift(56).over("item_id").alias("lag_56"),

    # Base series shifted by the complete forecast horizon
    pl.col("sales").shift(28).over("item_id").alias("sales_lag_base")
])

# Step 3: Calculate rolling averages from the shifted series
ca1_features = ca1_features.with_columns([
    pl.col("sales_lag_base")
    .rolling_mean(window_size=7)
    .over("item_id")
    .alias("rolling_mean_7"),

    pl.col("sales_lag_base")
    .rolling_mean(window_size=28)
    .over("item_id")
    .alias("rolling_mean_28"),

    pl.col("sales_lag_base")
    .rolling_mean(window_size=56)
    .over("item_id")
    .alias("rolling_mean_56")
])

# Step 4: Create calendar features
ca1_features = ca1_features.with_columns([
    pl.col("date").dt.weekday().alias("day_of_week"),
    pl.col("date").dt.month().alias("month"),

    pl.col("event_name_1")
    .fill_null("NoEvent")
    .alias("event_name_1")
])

# Step 5: Remove the temporary base feature
ca1_features = ca1_features.drop("sales_lag_base")

# Step 6: Collect the pilot store feature dataset
print("GENERATING FEATURES FOR:", PILOT_STORE)

ca1_features = ca1_features.collect()

# Step 7: Display dataset shape and preview
print("\nFEATURE DATASET SHAPE:", ca1_features.shape)

print("\nFEATURE DATASET PREVIEW")
display(ca1_features.head(10))

# Step 8: Check missing values in engineered features
feature_columns = [
    "lag_28",
    "lag_35",
    "lag_42",
    "lag_56",
    "rolling_mean_7",
    "rolling_mean_28",
    "rolling_mean_56"
]

print("\nFEATURE MISSING VALUE COUNTS")

feature_nulls = ca1_features.select([
    pl.col(column).null_count().alias(column)
    for column in feature_columns
])

display(feature_nulls)

# Step 9: Check missing selling prices
price_nulls = ca1_features.select(
    pl.col("sell_price").null_count()
).item()

print("\nMISSING SELLING PRICES:", price_nulls)

# Step 10: Verify the output
assert ca1_features.height == 5832737

print("\nFEATURE ENGINEERING COMPLETED FOR PILOT STORE!")
```

```python

# ============================================================
# CELL 20: LIGHTGBM L2 - CA_1 FOLD 1
# Purpose:
# 1. Prepare leak-free training and validation features.
# 2. Encode categorical features for LightGBM.
# 3. Train an L2 regression model.
# 4. Generate 28-day forecasts.
# 5. Evaluate MAE, RMSE and WAPE.
# 6. Save the trained model and predictions.
# ============================================================

import lightgbm as lgb
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Step 1: Define the Fold 1 date boundaries
fold1 = folds[0]

# Step 2: Define model features
categorical_features = [
    "item_id",
    "event_name_1",
    "day_of_week",
    "month"
]

numeric_features = [
    "sell_price",
    "snap_flag",
    "lag_28",
    "lag_35",
    "lag_42",
    "lag_56",
    "rolling_mean_7",
    "rolling_mean_28",
    "rolling_mean_56"
]

feature_columns = categorical_features + numeric_features

# Step 3: Prepare training data
# Exclude early rows where the 56-day rolling feature is unavailable.
train_pl = (
    ca1_features
    .filter(
        (pl.col("date") <= fold1["training_end"])
        & pl.col("rolling_mean_56").is_not_null()
    )
    .select(feature_columns + ["sales"])
)

# Step 4: Prepare validation data
validation_pl = (
    ca1_features
    .filter(
        pl.col("date").is_between(
            fold1["validation_start"],
            fold1["validation_end"],
            closed="both"
        )
    )
    .sort(["item_id", "date"])
)

# Step 5: Convert selected data to Pandas
train_pd = train_pl.to_pandas()
validation_pd = validation_pl.select(
    feature_columns + ["sales", "item_id", "store_id", "date"]
).to_pandas()

print("TRAINING ROWS:", len(train_pd))
print("VALIDATION ROWS:", len(validation_pd))

# Step 6: Convert numerical features to float32
for column in numeric_features:
    train_pd[column] = train_pd[column].astype("float32")
    validation_pd[column] = validation_pd[column].astype("float32")

# Step 7: Convert categorical features
for column in categorical_features:

    train_pd[column] = train_pd[column].astype("category")

    validation_pd[column] = pd.Categorical(
        validation_pd[column],
        categories=train_pd[column].cat.categories
    )

# Step 8: Create feature matrices and target arrays
X_train = train_pd[feature_columns]
y_train = train_pd["sales"].astype("float32")

X_validation = validation_pd[feature_columns]
y_validation = validation_pd["sales"].to_numpy(dtype=np.float32)

# Step 9: Initialize LightGBM L2 model
l2_model = lgb.LGBMRegressor(
    objective="regression_l2",
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    max_bin=127,
    min_child_samples=100,
    subsample=0.8,
    subsample_freq=1,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    n_jobs=4,
    random_state=42,
    verbosity=-1
)

# Step 10: Train with validation-based early stopping
print("\nSTARTING LIGHTGBM L2 TRAINING...")

l2_model.fit(
    X_train,
    y_train,
    categorical_feature=categorical_features,
    eval_set=[(X_validation, y_validation)],
    eval_metric="rmse",
    callbacks=[
        lgb.early_stopping(50),
        lgb.log_evaluation(50)
    ]
)

# Step 11: Generate validation predictions
l2_predictions = l2_model.predict(
    X_validation,
    num_iteration=l2_model.best_iteration_
)

# Sales cannot be negative, so clip negative forecasts
l2_predictions = np.maximum(l2_predictions, 0)

# Step 12: Calculate evaluation metrics
l2_mae = mean_absolute_error(
    y_validation,
    l2_predictions
)

l2_rmse = np.sqrt(
    mean_squared_error(y_validation, l2_predictions)
)

l2_wape = (
    np.sum(np.abs(y_validation - l2_predictions))
    / np.sum(y_validation)
) * 100

# Step 13: Display results
print("\nLIGHTGBM L2 - CA_1 FOLD 1 RESULTS")
print("Best iteration:", l2_model.best_iteration_)
print("MAE:", round(l2_mae, 4))
print("RMSE:", round(l2_rmse, 4))
print("WAPE (%):", round(l2_wape, 4))

# Step 14: Save the trained model
MODEL_DIR = PROJECT_DIR / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

l2_model.booster_.save_model(
    str(MODEL_DIR / "lightgbm_l2_ca1_fold1.txt")
)

# Step 15: Save predictions
l2_output = validation_pd[
    ["item_id", "store_id", "date", "sales"]
].copy()

l2_output = l2_output.rename(columns={
    "date": "forecast_date",
    "sales": "actual"
})

l2_output["horizon"] = (
    l2_output["forecast_date"] - fold1["validation_start"]
).dt.days + 1

l2_output["forecast"] = l2_predictions

l2_prediction_path = (
    PREDICTION_DIR / "lightgbm_l2_ca1_fold1.parquet"
)

pl.from_pandas(l2_output).write_parquet(
    l2_prediction_path,
    compression="zstd"
)

print("\nMODEL SAVED:", MODEL_DIR / "lightgbm_l2_ca1_fold1.txt")
print("PREDICTIONS SAVED:", l2_prediction_path)

print("\nLIGHTGBM L2 PILOT COMPLETED!")
```

```python

# ============================================================
# CELL 20: LIGHTGBM L2 REGRESSION - CA_1, FOLD 1
#
# Purpose:
# 1. Prepare training and validation datasets.
# 2. Select leak-free temporal features.
# 3. Handle categorical variables.
# 4. Train a LightGBM L2 regression model.
# 5. Generate 28-day sales forecasts.
# 6. Evaluate MAE, RMSE and WAPE.
# 7. Save the trained model and predictions.
# ============================================================

# Step 1: Import required libraries

import lightgbm as lgb
import pandas as pd
import numpy as np
import polars as pl

from sklearn.metrics import mean_absolute_error, mean_squared_error
from pathlib import Path

# Step 2: Define Fold 1 boundaries

fold1 = folds[0]

print("STORE:", PILOT_STORE)
print("TRAINING END:", fold1["training_end"])
print("VALIDATION START:", fold1["validation_start"])
print("VALIDATION END:", fold1["validation_end"])

# Step 3: Define model features

categorical_features = [
    "item_id",
    "event_name_1",
    "day_of_week",
    "month"
]

numeric_features = [
    "sell_price",
    "snap_flag",
    "lag_28",
    "lag_35",
    "lag_42",
    "lag_56",
    "rolling_mean_7",
    "rolling_mean_28",
    "rolling_mean_56"
]

feature_columns = categorical_features + numeric_features

# Step 4: Prepare training dataset
# Remove rows without the 56-day rolling feature.
# Preserve missing selling prices as null.

train_pl = (
    ca1_features
    .filter(
        (pl.col("date") <= fold1["training_end"])
        & pl.col("rolling_mean_56").is_not_null()
    )
    .select(feature_columns + ["sales"])
)

# Step 5: Prepare validation dataset
# Keep all validation rows, including the 12 missing-price rows.

validation_pl = (
    ca1_features
    .filter(
        pl.col("date").is_between(
            fold1["validation_start"],
            fold1["validation_end"],
            closed="both"
        )
    )
    .sort(["item_id", "date"])
)

# Step 6: Convert Polars datasets to Pandas
# Select each column only once to avoid DuplicateError.

train_pd = train_pl.to_pandas()

validation_pd = validation_pl.select(
    feature_columns + ["sales", "store_id", "date"]
).to_pandas()

print("\nDATASET SHAPES")
print("Training:", train_pd.shape)
print("Validation:", validation_pd.shape)

# Step 7: Convert numerical features to float32
# This reduces memory usage during model training.

for column in numeric_features:

    train_pd[column] = train_pd[column].astype("float32")

    validation_pd[column] = validation_pd[column].astype("float32")

# Step 8: Convert categorical features
# Use the training categories for validation as well.

for column in categorical_features:

    train_pd[column] = train_pd[column].astype("category")

    validation_pd[column] = pd.Categorical(
        validation_pd[column],
        categories=train_pd[column].cat.categories
    )

# Step 9: Prepare features and target values

X_train = train_pd[feature_columns]

y_train = train_pd["sales"].astype("float32")

X_validation = validation_pd[feature_columns]

y_validation = validation_pd["sales"].to_numpy(
    dtype=np.float32
)

print("\nTRAINING FEATURE SHAPE:", X_train.shape)
print("VALIDATION FEATURE SHAPE:", X_validation.shape)

# Step 10: Initialize LightGBM L2 model

l2_model = lgb.LGBMRegressor(
    objective="regression_l2",
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    max_bin=127,
    min_child_samples=100,
    subsample=0.8,
    subsample_freq=1,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    n_jobs=4,
    random_state=42,
    verbosity=-1
)

# Step 11: Train the model with early stopping

print("\nSTARTING LIGHTGBM L2 TRAINING...")

l2_model.fit(
    X_train,
    y_train,
    categorical_feature=categorical_features,
    eval_set=[
        (X_validation, y_validation)
    ],
    eval_metric="rmse",
    callbacks=[
        lgb.early_stopping(50),
        lgb.log_evaluation(50)
    ]
)

# Step 12: Generate validation predictions

l2_predictions = l2_model.predict(
    X_validation,
    num_iteration=l2_model.best_iteration_
)

# Sales forecasts cannot be negative.

l2_predictions = np.maximum(
    l2_predictions,
    0
)

# Step 13: Calculate evaluation metrics

l2_mae = mean_absolute_error(
    y_validation,
    l2_predictions
)

l2_rmse = np.sqrt(
    mean_squared_error(
        y_validation,
        l2_predictions
    )
)

l2_wape = (
    np.sum(np.abs(y_validation - l2_predictions))
    / np.sum(y_validation)
) * 100

# Step 14: Display evaluation results

print("\n======================================")
print("LIGHTGBM L2 - CA_1 FOLD 1 RESULTS")
print("======================================")

print("Best iteration:", l2_model.best_iteration_)

print("MAE:", round(l2_mae, 4))

print("RMSE:", round(l2_rmse, 4))

print("WAPE (%):", round(l2_wape, 4))

# Step 15: Save the trained model

MODEL_DIR = PROJECT_DIR / "models"

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

l2_model.booster_.save_model(
    str(MODEL_DIR / "lightgbm_l2_ca1_fold1.txt")
)

# Step 16: Prepare prediction output

l2_output = validation_pd[
    ["item_id", "store_id", "date", "sales"]
].copy()

l2_output = l2_output.rename(
    columns={
        "date": "forecast_date",
        "sales": "actual"
    }
)

l2_output["horizon"] = (
    l2_output["forecast_date"] - fold1["validation_start"]
).dt.days + 1

l2_output["forecast"] = l2_predictions

# Step 17: Save predictions as Parquet

PREDICTION_DIR.mkdir(
    parents=True,
    exist_ok=True
)

l2_prediction_path = (
    PREDICTION_DIR / "lightgbm_l2_ca1_fold1.parquet"
)

pl.from_pandas(l2_output).write_parquet(
    l2_prediction_path,
    compression="zstd"
)

# Step 18: Verify saved prediction count

assert len(l2_output) == EXPECTED_SERIES_PER_DAY * HORIZON

assert l2_output["forecast"].notna().all()

print("\n======================================")
print("SAVED OUTPUTS")
print("======================================")

print("MODEL:", MODEL_DIR / "lightgbm_l2_ca1_fold1.txt")

print("PREDICTIONS:", l2_prediction_path)

print("PREDICTION ROWS:", len(l2_output))

print("\nLIGHTGBM L2 PILOT COMPLETED!")
```

```python

# ============================================================
# CELL 22: LIGHTGBM QUANTILE REGRESSION - CA_1, FOLD 1
# Purpose:
# 1. Train a lower quantile model (0.05).
# 2. Train an upper quantile model (0.95).
# 3. Generate prediction intervals for 28 days.
# 4. Calculate pinball loss and interval coverage.
# 5. Save both models and their predictions.
# ============================================================

# Step 1: Define quantile levels

QUANTILES = [0.05, 0.95]

quantile_models = {}
quantile_predictions = {}

# Step 2: Train each quantile model

for alpha in QUANTILES:

    print(f"\nTRAINING QUANTILE MODEL: {alpha}")

    quantile_model = lgb.LGBMRegressor(
        objective="quantile",
        alpha=alpha,
        n_estimators=500,
        learning_rate=0.05,
        num_leaves=31,
        max_bin=127,
        min_child_samples=100,
        subsample=0.8,
        subsample_freq=1,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        n_jobs=4,
        random_state=42,
        verbosity=-1
    )

    # Train the model using the same data and features
    quantile_model.fit(
        X_train,
        y_train,
        categorical_feature=categorical_features,
        eval_set=[
            (X_validation, y_validation)
        ],
        eval_metric="quantile",
        callbacks=[
            lgb.early_stopping(50),
            lgb.log_evaluation(50)
        ]
    )

    # Generate validation predictions
    predictions = quantile_model.predict(
        X_validation,
        num_iteration=quantile_model.best_iteration_
    )

    # Save model and predictions in memory
    quantile_models[alpha] = quantile_model
    quantile_predictions[alpha] = predictions

    # Save the trained model
    quantile_model_path = (
        MODEL_DIR / f"lightgbm_quantile_{alpha}_ca1_fold1.txt"
    )

    quantile_model.booster_.save_model(
        str(quantile_model_path)
    )

    print("Best iteration:", quantile_model.best_iteration_)
    print("Model saved:", quantile_model_path)

# Step 3: Extract lower and upper predictions

raw_lower = quantile_predictions[0.05]
raw_upper = quantile_predictions[0.95]

# Step 4: Count negative predictions and quantile crossings

negative_lower_count = int(np.sum(raw_lower < 0))
negative_upper_count = int(np.sum(raw_upper < 0))

crossing_count = int(np.sum(raw_lower > raw_upper))

print("\nQUANTILE DIAGNOSTICS")
print("Negative lower predictions:", negative_lower_count)
print("Negative upper predictions:", negative_upper_count)
print("Crossed intervals:", crossing_count)

# Step 5: Ensure non-negative and ordered intervals

lower_predictions = np.maximum(
    np.minimum(raw_lower, raw_upper),
    0
)

upper_predictions = np.maximum(
    np.maximum(raw_lower, raw_upper),
    0
)

# Step 6: Calculate pinball loss

def calculate_pinball_loss(actual, predicted, alpha):

    errors = actual - predicted

    return np.mean(
        np.maximum(
            alpha * errors,
            (alpha - 1) * errors
        )
    )

lower_pinball = calculate_pinball_loss(
    y_validation,
    lower_predictions,
    0.05
)

upper_pinball = calculate_pinball_loss(
    y_validation,
    upper_predictions,
    0.95
)

# Step 7: Calculate raw interval coverage and width

interval_coverage = np.mean(
    (y_validation >= lower_predictions)
    & (y_validation <= upper_predictions)
) * 100

interval_width = np.mean(
    upper_predictions - lower_predictions
)

# Step 8: Display results

print("\nQUANTILE FORECAST RESULTS")
print("Lower pinball loss:", round(lower_pinball, 4))
print("Upper pinball loss:", round(upper_pinball, 4))
print("Interval coverage (%):", round(interval_coverage, 4))
print("Average interval width:", round(interval_width, 4))

# Step 9: Create prediction output

quantile_output = validation_pd[
    ["item_id", "store_id", "date", "sales"]
].copy()

quantile_output = quantile_output.rename(
    columns={
        "date": "forecast_date",
        "sales": "actual"
    }
)

quantile_output["forecast_date"] = pd.to_datetime(
    quantile_output["forecast_date"]
)

quantile_output["horizon"] = (
    quantile_output["forecast_date"]
    - pd.Timestamp(fold1["validation_start"])
).dt.days + 1

quantile_output["lower_05"] = lower_predictions
quantile_output["upper_95"] = upper_predictions

# Step 10: Save quantile predictions

quantile_prediction_path = (
    PREDICTION_DIR / "quantile_predictions_ca1_fold1.parquet"
)

pl.from_pandas(quantile_output).write_parquet(
    quantile_prediction_path,
    compression="zstd"
)

# Step 11: Verify output

assert len(quantile_output) == 85372
assert (quantile_output["lower_05"] <= quantile_output["upper_95"]).all()

print("\nPREDICTION ROWS:", len(quantile_output))
print("PREDICTIONS SAVED:", quantile_prediction_path)

print("\nQUANTILE MODEL IMPLEMENTATION COMPLETED!")
```

```python

# ============================================================
# CELL 26: GENERATE FOLD 2 AND FOLD 3 QUANTILE PREDICTIONS
# Purpose:
# 1. Prepare validation features for CA_1 in Folds 2 and 3.
# 2. Preserve the exact categorical levels from Fold 1.
# 3. Generate lower (0.05) and upper (0.95) predictions.
# 4. Save predictions for conformal calibration and testing.
# ============================================================

# Store predictions separately for each fold.
fold_quantile_outputs = {}

# Process Fold 2 and Fold 3.
for fold_number in [2, 3]:

    print(f"\n{'=' * 55}")
    print(f"GENERATING QUANTILE PREDICTIONS: FOLD {fold_number}")
    print(f"{'=' * 55}")

    # Get the validation date boundaries for CA_1.
    validation_summary = (
        fold_splits[fold_number]["validation"]
        .filter(pl.col("store_id") == PILOT_STORE)
        .select([
            pl.col("date").min().alias("start_date"),
            pl.col("date").max().alias("end_date")
        ])
        .collect()
    )

    start_date = validation_summary["start_date"][0]
    end_date = validation_summary["end_date"][0]

    # Extract the engineered features for this validation period.
    fold_data = (
        ca1_features
        .filter(
            (pl.col("store_id") == PILOT_STORE) &
            (pl.col("date") >= start_date) &
            (pl.col("date") <= end_date)
        )
        .sort(["item_id", "date"])
        .to_pandas()
    )

    print("Validation period:", start_date, "to", end_date)
    print("Validation rows:", len(fold_data))

    # Prepare the model input using the same feature order.
    X_fold = fold_data[feature_columns].copy()

    # Preserve the categorical levels used during Fold 1 training.
    for column in ["item_id", "event_name_1",
                   "day_of_week", "month"]:

        category_type = pd.CategoricalDtype(
            categories=X_train[column].cat.categories
        )

        X_fold[column] = X_fold[column].astype(category_type)

    # Convert numerical columns to the training data's float32 type.
    numeric_columns = [
        column for column in feature_columns
        if column not in [
            "item_id", "event_name_1", "day_of_week", "month"
        ]
    ]

    for column in numeric_columns:
        X_fold[column] = X_fold[column].astype("float32")

    # Generate lower and upper quantile predictions.
    raw_lower = quantile_models[0.05].predict(
        X_fold,
        num_iteration=quantile_models[0.05].best_iteration_
    )

    raw_upper = quantile_models[0.95].predict(
        X_fold,
        num_iteration=quantile_models[0.95].best_iteration_
    )

    # Count any quantile crossing before correction.
    crossing_count = np.sum(raw_lower > raw_upper)

    # Ensure lower and upper boundaries are correctly ordered
    # and that predicted sales cannot be negative.
    lower_predictions = np.maximum(
        np.minimum(raw_lower, raw_upper), 0
    )

    upper_predictions = np.maximum(
        np.maximum(raw_lower, raw_upper), 0
    )

    # Create the prediction output.
    fold_output = pd.DataFrame({
        "item_id": fold_data["item_id"].astype(str),
        "store_id": fold_data["store_id"].astype(str),
        "forecast_date": pd.to_datetime(fold_data["date"]),
        "actual": fold_data["sales"].astype("float32"),
        "lower_05": lower_predictions.astype("float32"),
        "upper_95": upper_predictions.astype("float32")
    })

    # Calculate forecast horizon (1 to 28).
    fold_output["horizon"] = (
        (fold_output["forecast_date"] - pd.Timestamp(start_date))
        .dt.days + 1
    )

    # Verify prediction results.
    print("\nPREDICTION CHECKS")
    print("Rows:", len(fold_output))
    print("Crossed raw intervals:", crossing_count)
    print("Null lower predictions:", fold_output["lower_05"].isna().sum())
    print("Null upper predictions:", fold_output["upper_95"].isna().sum())
    print("Minimum lower prediction:", fold_output["lower_05"].min())
    print("Maximum upper prediction:", fold_output["upper_95"].max())
    print("Minimum horizon:", fold_output["horizon"].min())
    print("Maximum horizon:", fold_output["horizon"].max())

    # Save the predictions.
    output_path = (
        PROJECT_DIR / "predictions" /
        f"quantile_predictions_ca1_fold{fold_number}.parquet"
    )

    fold_output.to_parquet(output_path, index=False)

    fold_quantile_outputs[fold_number] = fold_output

    print("\nSaved:", output_path)

print("\nFOLD 2 AND FOLD 3 QUANTILE PREDICTIONS COMPLETED!")
```

```python

# ============================================================
# CELL 27: CONFORMAL CALIBRATION AND FOLD 3 EVALUATION
# Purpose:
# 1. Calculate conformity scores using Fold 2.
# 2. Find the finite-sample 90% conformal adjustment.
# 3. Apply the adjustment to Fold 3 prediction intervals.
# 4. Compare raw and calibrated coverage and interval width.
# 5. Save the calibrated Fold 3 predictions.
# ============================================================

# Set the target miscoverage.
# 0.10 means we want a nominal 90% prediction interval.
ALPHA = 0.10

# Retrieve Fold 2 calibration predictions and Fold 3 test predictions.
calibration_data = fold_quantile_outputs[2].copy()
test_data = fold_quantile_outputs[3].copy()

# ------------------------------------------------------------
# STEP 1: Calculate Fold 2 conformity scores
# ------------------------------------------------------------

# A score of zero means actual sales are already inside the interval.
# A positive score measures how far actual sales fall outside it.

calibration_scores = np.maximum(
    calibration_data["lower_05"].to_numpy() -
    calibration_data["actual"].to_numpy(),

    calibration_data["actual"].to_numpy() -
    calibration_data["upper_95"].to_numpy()
)

calibration_scores = np.maximum(calibration_scores, 0)

# ------------------------------------------------------------
# STEP 2: Calculate the finite-sample conformal adjustment
# ------------------------------------------------------------

n_calibration = len(calibration_scores)

# Finite-sample conformal rank.
conformal_rank = int(
    np.ceil((n_calibration + 1) * (1 - ALPHA))
)

# Sort the scores and select the required rank.
sorted_scores = np.sort(calibration_scores)

# Convert the one-based rank to a zero-based Python index.
conformal_index = min(conformal_rank - 1, n_calibration - 1)

conformal_adjustment = sorted_scores[conformal_index]

print("CONFORMAL CALIBRATION RESULTS")
print("Calibration observations:", n_calibration)
print("Target coverage:", (1 - ALPHA) * 100, "%")
print("Conformal rank:", conformal_rank)
print("Conformal adjustment:", conformal_adjustment)

# ------------------------------------------------------------
# STEP 3: Measure raw Fold 2 calibration coverage
# ------------------------------------------------------------

fold2_covered = (
    (calibration_data["actual"] >= calibration_data["lower_05"]) &
    (calibration_data["actual"] <= calibration_data["upper_95"])
)

fold2_raw_coverage = fold2_covered.mean() * 100

# Expand Fold 2 intervals using the conformal adjustment.
fold2_calibrated_lower = np.maximum(
    calibration_data["lower_05"] - conformal_adjustment, 0
)

fold2_calibrated_upper = (
    calibration_data["upper_95"] + conformal_adjustment
)

fold2_calibrated_coverage = (
    (calibration_data["actual"] >= fold2_calibrated_lower) &
    (calibration_data["actual"] <= fold2_calibrated_upper)
).mean() * 100

print("\nFOLD 2 COVERAGE")
print("Raw coverage:", round(fold2_raw_coverage, 4), "%")
print("Calibrated coverage:", round(fold2_calibrated_coverage, 4), "%")

# ------------------------------------------------------------
# STEP 4: Evaluate raw Fold 3 intervals
# ------------------------------------------------------------

fold3_actual = test_data["actual"].to_numpy()
fold3_lower = test_data["lower_05"].to_numpy()
fold3_upper = test_data["upper_95"].to_numpy()

fold3_raw_coverage = (
    (fold3_actual >= fold3_lower) &
    (fold3_actual <= fold3_upper)
).mean() * 100

fold3_raw_width = np.mean(fold3_upper - fold3_lower)

# ------------------------------------------------------------
# STEP 5: Apply Fold 2 adjustment to Fold 3
# ------------------------------------------------------------

test_data["calibrated_lower"] = np.maximum(
    fold3_lower - conformal_adjustment, 0
)

test_data["calibrated_upper"] = (
    fold3_upper + conformal_adjustment
)

# Calculate calibrated Fold 3 coverage.
fold3_calibrated_coverage = (
    (fold3_actual >= test_data["calibrated_lower"]) &
    (fold3_actual <= test_data["calibrated_upper"])
).mean() * 100

# Calculate calibrated average interval width.
fold3_calibrated_width = np.mean(
    test_data["calibrated_upper"] -
    test_data["calibrated_lower"]
)

# ------------------------------------------------------------
# STEP 6: Display final comparison
# ------------------------------------------------------------

print("\nFOLD 3 FINAL EVALUATION")
print("------------------------------------------")
print("Raw coverage (%):",
      round(fold3_raw_coverage, 4))

print("Calibrated coverage (%):",
      round(fold3_calibrated_coverage, 4))

print("Target coverage (%):",
      (1 - ALPHA) * 100)

print("Raw average width:",
      round(fold3_raw_width, 4))

print("Calibrated average width:",
      round(fold3_calibrated_width, 4))

print("Coverage improvement (percentage points):",
      round(fold3_calibrated_coverage - fold3_raw_coverage, 4))

# ------------------------------------------------------------
# STEP 7: Save calibrated Fold 3 predictions
# ------------------------------------------------------------

calibrated_output_path = (
    PROJECT_DIR / "predictions" /
    "conformal_predictions_ca1_fold3.parquet"
)

test_data.to_parquet(calibrated_output_path, index=False)

print("\nCALIBRATED PREDICTIONS SAVED:")
print(calibrated_output_path)

print("\nCONFORMAL CALIBRATION COMPLETED!")
```

```python

# ============================================================
# CELL 28: TEMPORAL COVERAGE AND INTERVAL WIDTH ANALYSIS
# Purpose:
# 1. Compare Fold 2 and Fold 3 coverage.
# 2. Calculate coverage for each of the 28 forecast horizons.
# 3. Examine interval width across forecast horizons.
# 4. Identify whether prediction uncertainty changes over time.
# ============================================================

# Create copies of the two prediction datasets.
fold2_check = fold_quantile_outputs[2].copy()
fold3_check = fold_quantile_outputs[3].copy()

# Calculate coverage and interval width for each fold.
for fold_number, data in [
    (2, fold2_check),
    (3, fold3_check)
]:

    data["covered"] = (
        (data["actual"] >= data["lower_05"]) &
        (data["actual"] <= data["upper_95"])
    )

    data["interval_width"] = (
        data["upper_95"] - data["lower_05"]
    )

    print(f"\nFOLD {fold_number} SUMMARY")
    print("--------------------------------")

    print("Date range:",
          data["forecast_date"].min(),
          "to",
          data["forecast_date"].max())

    print("Total observations:", len(data))
    print("Covered observations:", int(data["covered"].sum()))
    print("Coverage (%):", round(data["covered"].mean() * 100, 4))
    print("Average interval width:",
          round(data["interval_width"].mean(), 4))

# ------------------------------------------------------------
# Calculate coverage separately for every forecast horizon.
# ------------------------------------------------------------

horizon_results = []

for fold_number, data in [
    (2, fold2_check),
    (3, fold3_check)
]:

    for horizon, group in data.groupby("horizon"):

        horizon_results.append({
            "fold": fold_number,
            "horizon": int(horizon),
            "observations": len(group),
            "covered_count": int(group["covered"].sum()),
            "coverage_pct": group["covered"].mean() * 100,
            "average_interval_width": group["interval_width"].mean()
        })

horizon_df = pd.DataFrame(horizon_results)

print("\nCOVERAGE BY FORECAST HORIZON")
print("--------------------------------")

display(
    horizon_df.round(4)
)

# ------------------------------------------------------------
# Compare the distribution of conformity scores.
# ------------------------------------------------------------

print("\nFOLD 2 CONFORMITY SCORE DIAGNOSTICS")
print("--------------------------------")

print("Zero-score percentage:",
      round((calibration_scores == 0).mean() * 100, 4))

print("Score 90th percentile:",
      np.quantile(calibration_scores, 0.90))

print("Score 95th percentile:",
      np.quantile(calibration_scores, 0.95))

print("Maximum score:",
      calibration_scores.max())

# Save horizon-level results.
horizon_path = (
    PROJECT_DIR / "results" /
    "conformal_horizon_coverage_ca1.csv"
)

horizon_df.to_csv(horizon_path, index=False)

print("\nHORIZON ANALYSIS SAVED:")
print(horizon_path)
```

```python

# ============================================================
# CELL 30: HORIZON-SPECIFIC CONFORMAL CALIBRATION
# Purpose:
# 1. Calculate conformity scores separately for each horizon.
# 2. Find an independent adjustment for each of the 28 days.
# 3. Apply these adjustments to Fold 3.
# 4. Compare raw and calibrated coverage and interval widths.
# ============================================================

ALPHA = 0.10

# Use Fold 2 for calibration and Fold 3 for evaluation.
calibration_data = fold_quantile_outputs[2].copy()
test_data = fold_quantile_outputs[3].copy()

temporal_results = []

# Create columns for horizon-calibrated intervals.
test_data["temporal_lower"] = np.nan
test_data["temporal_upper"] = np.nan

# Process each forecast horizon independently.
for horizon in range(1, HORIZON + 1):

    # --------------------------------------------------------
    # STEP 1: Select the calibration observations for this day.
    # --------------------------------------------------------

    calibration_group = calibration_data[
        calibration_data["horizon"] == horizon
    ]

    test_group = test_data[
        test_data["horizon"] == horizon
    ]

    # --------------------------------------------------------
    # STEP 2: Calculate conformity scores.
    # --------------------------------------------------------

    scores = np.maximum(
        calibration_group["lower_05"].to_numpy()
        - calibration_group["actual"].to_numpy(),

        calibration_group["actual"].to_numpy()
        - calibration_group["upper_95"].to_numpy()
    )

    scores = np.maximum(scores, 0)

    # --------------------------------------------------------
    # STEP 3: Calculate finite-sample adjustment.
    # --------------------------------------------------------

    n = len(scores)

    rank = int(np.ceil((n + 1) * (1 - ALPHA)))

    sorted_scores = np.sort(scores)

    adjustment = sorted_scores[
        min(rank - 1, n - 1)
    ]

    # --------------------------------------------------------
    # STEP 4: Apply adjustment to Fold 3.
    # --------------------------------------------------------

    test_indices = test_group.index

    calibrated_lower = np.maximum(
        test_group["lower_05"].to_numpy() - adjustment,
        0
    )

    calibrated_upper = (
        test_group["upper_95"].to_numpy() + adjustment
    )

    test_data.loc[test_indices, "temporal_lower"] = calibrated_lower
    test_data.loc[test_indices, "temporal_upper"] = calibrated_upper

    # --------------------------------------------------------
    # STEP 5: Evaluate Fold 3 coverage for this horizon.
    # --------------------------------------------------------

    actual = test_group["actual"].to_numpy()

    raw_coverage = np.mean(
        (actual >= test_group["lower_05"].to_numpy()) &
        (actual <= test_group["upper_95"].to_numpy())
    ) * 100

    calibrated_coverage = np.mean(
        (actual >= calibrated_lower) &
        (actual <= calibrated_upper)
    ) * 100

    raw_width = np.mean(
        test_group["upper_95"].to_numpy() -
        test_group["lower_05"].to_numpy()
    )

    calibrated_width = np.mean(
        calibrated_upper - calibrated_lower
    )

    # Save the results for this horizon.
    temporal_results.append({
        "horizon": horizon,
        "calibration_rows": n,
        "conformal_adjustment": adjustment,
        "fold3_raw_coverage_pct": raw_coverage,
        "fold3_calibrated_coverage_pct": calibrated_coverage,
        "raw_interval_width": raw_width,
        "calibrated_interval_width": calibrated_width
    })

# ------------------------------------------------------------
# STEP 6: Display temporal calibration results.
# ------------------------------------------------------------

temporal_calibration_df = pd.DataFrame(temporal_results)

print("HORIZON-SPECIFIC CONFORMAL CALIBRATION")
print("------------------------------------------")

display(temporal_calibration_df.round(4))

# ------------------------------------------------------------
# STEP 7: Calculate overall results.
# ------------------------------------------------------------

raw_overall_coverage = np.mean(
    (test_data["actual"] >= test_data["lower_05"]) &
    (test_data["actual"] <= test_data["upper_95"])
) * 100

calibrated_overall_coverage = np.mean(
    (test_data["actual"] >= test_data["temporal_lower"]) &
    (test_data["actual"] <= test_data["temporal_upper"])
) * 100

raw_overall_width = np.mean(
    test_data["upper_95"] - test_data["lower_05"]
)

calibrated_overall_width = np.mean(
    test_data["temporal_upper"] - test_data["temporal_lower"]
)

print("\nOVERALL TEMPORAL CALIBRATION RESULTS")
print("------------------------------------------")
print("Raw coverage:", round(raw_overall_coverage, 4), "%")
print("Calibrated coverage:", round(calibrated_overall_coverage, 4), "%")
print("Target coverage:", (1 - ALPHA) * 100, "%")
print("Raw interval width:", round(raw_overall_width, 4))
print("Calibrated interval width:", round(calibrated_overall_width, 4))

print("\nNon-zero horizon adjustments:",
      int((temporal_calibration_df["conformal_adjustment"] > 0).sum()))

# ------------------------------------------------------------
# STEP 8: Save temporal calibration outputs.
# ------------------------------------------------------------

temporal_results_path = (
    PROJECT_DIR / "results" /
    "horizon_specific_calibration_ca1.csv"
)

temporal_predictions_path = (
    PROJECT_DIR / "predictions" /
    "temporal_calibrated_predictions_ca1_fold3.parquet"
)

temporal_calibration_df.to_csv(
    temporal_results_path,
    index=False
)

test_data.to_parquet(
    temporal_predictions_path,
    index=False
)

print("\nSAVED RESULTS:")
print(temporal_results_path)
print(temporal_predictions_path)

print("\nHORIZON-SPECIFIC CALIBRATION COMPLETED!")
```
